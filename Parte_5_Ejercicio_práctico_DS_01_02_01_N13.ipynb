`````python
def buscar_elemento(valor, estructura):
    """Comprueba si un valor existe en una tupla o lista."""
    return valor in estructura

ejer_9 = (
    3,
    20,
    3,
    47,
    19,
    3,
    29,
    45,
    67,
    78,
    90,
    3,
    3,
    5,
    2,
    4,
    7,
    9,
    4,
    2,
    4,
    3,
    3,
    4,
    6,
    7,
)

print(
    "¿Está el 60 en la tupla?:", buscar_elemento(60, ejer_9)
)
print("¿Está el 47 en la tupla?:", buscar_elemento(47, ejer_9))

print("\n" + "=" * 50 + "\n")

ejer_3 = [
    "Un",
    "árbol",
    "binario",
    "es",
    "una",
    "estructura",
    "de",
    "un",
    "tipo",
    "particular",
    "a",
    "veces",
    "no",
    "es",
    "ni",
    "binario",
]


def obtener_indices_duplicados(duplicado, lista, case_sensitive=False):
    """Devuelve una lista con todos los índices en los que aparece el elemento 'duplicado'."""
    indices = []
    for i, elemento in enumerate(lista):
        if case_sensitive:
            coincide = elemento == duplicado
        else:
                        coincide = str(elemento).lower() == str(duplicado).lower()

        if coincide:
            indices.append(i)

    return indices


elementos_a_buscar = ["un", "es", "binario"]

for elem in elementos_a_buscar:
    indices = obtener_indices_duplicados(elem, ejer_3)
    print(
        f"Índices de '{elem}': {indices} (Aparece {len(indices)} veces en las posiciones {[ejer_3[i] for i in indices]})"
    )
`````

## Ejercicio 11
Convierte el programa del ejercicio 9 en una función al que se le pasen como argumentos la librería, posicional, y el título, argumento keyword con valor "ninguno", por defecto.

`````python
def buscar_libro(libreria, *, titulo="ninguno"):
    """Busca libros por título dentro de la librería recibida.

    - libreria: Argumento posicional con la lista de libros.
    - titulo: Argumento keyword-only (por defecto "ninguno").
    """
    if titulo == "ninguno":
        return "Debes indicar un título para buscar."

    libros_encontrados = [
        libro
        for libro in libreria
        if libro["titulo"].lower() == titulo.lower()
    ]

    return libros_encontrados if libros_encontrados else "Ese no lo tengo, ¿mola?"


libreria = [
    {
        "titulo": "Cien años de soledad",
        "autor": "Gabriel García Márquez",
        "idioma original": "esperanto",
        "año de publicación": 1967,
    },
    {
        "titulo": "1984",
        "autor": "George Orwell",
        "idioma original": "esperanto",
        "año de publicación": 1949,
    },
    {
        "titulo": "El Principito",
        "autor": "Antoine de Saint-Exupéry",
        "idioma original": "esperanto",
        "año de publicación": 1943,
    },
    {
        "titulo": "Tokio blues",
        "autor": "Haruki Murakami",
        "idioma original": "esperanto",
        "año de publicación": 1987,
    },
]


resultado_1 = buscar_libro(libreria, titulo="1984")
print("Búsqueda 1 ('1984'):")
print(resultado_1)

print("\n" + "=" * 50 + "\n")

resultado_2 = buscar_libro(libreria, titulo="Don Quijote de la Mancha")
print("Búsqueda 2 ('Don Quijote de la Mancha'):")
print(resultado_2)

print("\n" + "=" * 50 + "\n")

resultado_3 = buscar_libro(libreria)
print("Búsqueda 3 (sin argumento titulo):")
print(resultado_3)
`````

## Ejercicio 13
Escribe una función que compruebe si un DNI es válido, devolviendo `True` en caso afirmativo.  
En primer lugar, el DNI debe tener entre 7 y 8 dígitos numéricos.  
Para saber si la letra del DNI es correcta puedes utilizar el siguiente fragmento de código:
>```Python
>DNI= 55555555
>palabra='TRWAGMYFPDXBNJZSQVHLCKE'
>print('La letra del DNI es: ', palabra[DNI%23])
>````


`````python
def es_dni_valido(dni):
    """Comprueba si un DNI introducido como string es válido.

    Devuelve True si el formato y la letra son correctos, False en caso
    contrario.
    """
    LETRAS_DNI = "TRWAGMYFPDXBNJZSQVHLCKE"

    dni = str(dni).strip()

    numeros = dni[:-1]
    letra = dni[-1].upper()

    if not numeros.isdigit() or not (7 <= len(numeros) <= 8):
        return False

    if not letra.isalpha():
        return False

    resto = int(numeros) % 23
    letra_calculada = LETRAS_DNI[resto]

    return letra == letra_calculada


dnis_de_prueba = [
    "12345678Z",  # Ejemplo (inválido por letra)
    "12345678Z",  # Formato correcto
    "1234567A",  # 7 dígitos + letra
    "123456789A",  # Inválido (9 dígitos)
    "ABC12345Z",  # Inválido (contiene letras en el número)
]

dni_usuario = input("Introduce un DNI para validar: ")

if es_dni_valido(dni_usuario):
    print(f"El DNI {dni_usuario} es VÁLIDO.")
else:
    print(f"El DNI {dni_usuario} NO es válido.")
`````
