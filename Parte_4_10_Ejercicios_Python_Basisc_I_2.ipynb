## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

radianes = grados*(pi/180)

```python
import math
grados = float(input("Introduce los grados a convertir: "))
radianes = grados * (math.pi / 180)
print(f"{grados}° equivalen a {radianes:.4f} radianes.")
```

## Ejercicio 4
Escribe un programa que calcule el area de un paralelogramo (base x altura). También con `input`. En este caso debe ser capaz de admitir valores reales, con decimales, para base y altura en español. (PISTA: Recuerda que tendrás que hacer uso del método replace)

```python
entrada_base = input("Introduce la base del paralelogramo (p. ej. 10,5): ")
entrada_altura = input("Introduce la altura del paralelogramo (p. ej. 4,2): ")
base = float(entrada_base.replace(",", "."))
altura = float(entrada_altura.replace(",", "."))
area = base * altura
area_formateada = str(round(area, 2)).replace(".", ",")
print(f"El área del paralelogramo es: {area_formateada}")
```

## Ejercicio 5
Complete el siguiente código para solicitar al usuario su nombre y su edad. Luego, calcule en qué año nació el usuario y muestre el resultado.

```python
import datetime

nombre = input("Introduce tu nombre: ")
edad = int(input("Introduce tu edad: "))
año_actual = datetime.datetime.now().year
año_nacimiento = año_actual - edad
print(f"Hola {nombre}, naciste en el año {año_nacimiento} (o {año_nacimiento - 1} si aún no has cumplido años este año).")
```
