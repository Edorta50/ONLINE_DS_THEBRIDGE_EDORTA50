## Ejercicio 1
Somos una empresa que tiene varias tiendas de electrodomesticos. Necesitamos un programa para manejar las tiendas, ver las ventas que han tenido, cuántos empleados hay, etc... Para ello vamos a modelizar la tienda en una clase, que tendrá las siguientes características:
* Nombre clase: "Tienda"
* Atributos comunes:
    * Tipo: "Electrodomésticos"
    * Abierta: True
* Atributos propios de cada tienda:
    * Nombre - String
    * Dirección - String
    * Número de empleados - int
    * Ventas ultimos 3 meses - Lista de 3 numeros
* Método para calcular las ventas de todos los meses, que devuelva un numero con todas las ventas.
* Método que calcula la media de ventas de los ultimos meses, por empleado
* Método que devuelve en un string el nombre de la tienda, junto con su dirección.
* Método que obtiene las ventas del último mes.
* Método para dar una proyección de las ventas en caso de haber invertido X dinero en marketing en esa tienda. Siendo X el parámetro de entrada. Si X es menor de 1000, las ventas de los ultimos 3 meses hubiesen sido de (1.2 x ventas), si es mayor o igual de 1000, las ventas hubiesen sido de (1.5 x venta). El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve las ventas con el aumento.

Se pide:
1. Implementar la clase "Tienda"
2. Crear tres tiendas con datos inventados
3. Comprueba en al menos una de ellas, todo lo implementado en la clase tienda (sus atributos, media de ventas, ventas/empleado...)
3. Calcular las ventas del último mes de todas las tiendas. Para ello usa el bucle `for`
4. Imprime por pantalla los nombres de las tiendas cuya dirección lleve el string "Avenida"
5. Seguro que ya has hecho este apartado... Documenta la clase :)

```python
class Tienda:

    tipo = "Electrodomésticos"
    abierta = True

    def __init__(
        self,
        nombre: str,
        direccion: str,
        numero_empleados: int,
        ventas_ultimos_3_meses: list[float],
    ):
        """Initializes a Tienda object.

        Args:
            nombre (str): Nombre comercial de la tienda.
            direccion (str): Dirección física de la tienda.
            numero_empleados (int): Cantidad de empleados contratados.
            ventas_ultimos_3_meses (list): Lista con el importe de ventas de los
              últimos 3 meses.
        """
        self.nombre = nombre
        self.direccion = direccion
        self.numero_empleados = numero_empleados
        self.ventas_ultimos_3_meses = ventas_ultimos_3_meses

    def total_ventas(self) -> float:
        """Calcula la suma total de las ventas de los últimos tres meses.

        Returns:
            float: Suma total de las ventas.
        """
        return sum(self.ventas_ultimos_3_meses)

    def media_ventas_por_empleado(self) -> float:
        """Calcula la media de ventas acumuladas por empleado.

        Returns:
            float: Promedio de ventas atribuible a cada empleado.
            Devuelve 0.0 si la tienda no tiene empleados para evitar división por cero.
        """
        if self.numero_empleados == 0:
            return 0.0
        return self.total_ventas() / self.numero_empleados

    def obtener_informacion_ubicacion(self) -> str:
        """Devuelve el nombre de la tienda y su dirección en un solo string.

        Returns:
            str: Cadena formateada con el nombre y la dirección.
        """
        return f"{self.nombre} - {self.direccion}"

    def ventas_ultimo_mes(self) -> float:
        """Obtiene el importe facturado en el último mes registrado.

        Returns:
            float: Ventas correspondientes al último elemento de la lista.
        """
        return self.ventas_ultimos_3_meses[-1]

    def proyeccion_marketing(self, x: float) -> float:
        """Recalcula las ventas aplicando un factor de crecimiento según la inversión en marketing.

        Actualiza la lista de ventas de los últimos 3 meses con la proyección calculada
        y devuelve el total acumulado resultante.

        Args:
            x (float): Cantidad invertida en marketing.

        Returns:
            float: Suma total de las ventas proyectadas.
        """
        factor = 1.2 if x < 1000 else 1.5
        self.ventas_ultimos_3_meses = [
            venta * factor for venta in self.ventas_ultimos_3_meses
        ]
        return self.total_ventas()


tienda1 = Tienda(
    nombre="ElectroSur Centro",
    direccion="Avenida de la Constitución 45",
    numero_empleados=8,
    ventas_ultimos_3_meses=[12000.0, 15000.0, 18000.0],
)

tienda2 = Tienda(
    nombre="ElectroSur Norte",
    direccion="Calle Mayor 12",
    numero_empleados=5,
    ventas_ultimos_3_meses=[8500.0, 9200.0, 11000.0],
)

tienda3 = Tienda(
    nombre="ElectroSur Exprés",
    direccion="Avenida de América 102",
    numero_empleados=4,
    ventas_ultimos_3_meses=[6000.0, 7500.0, 8000.0],
)

lista_tiendas = [tienda1, tienda2, tienda3]


print("--- COMPROBACIÓN DE FUNCIONALIDADES (Tienda 1) ---")
print(f"Tipo (Atributo de clase): {tienda1.tipo}")
print(f"Abierta (Atributo de clase): {tienda1.abierta}")
print(f"Ubicación: {tienda1.obtener_informacion_ubicacion()}")
print(f"Número de empleados: {tienda1.numero_empleados}")
print(f"Ventas actuales (3 meses): {tienda1.ventas_ultimos_3_meses} €")
print(f"Total ventas: {tienda1.total_ventas():.2f} €")
print(
    f"Media ventas por empleado: {tienda1.media_ventas_por_empleado():.2f} €"
)
print(f"Ventas del último mes: {tienda1.ventas_ultimo_mes():.2f} €")

total_proyectado = tienda1.proyeccion_marketing(1500)
print(
    f"\n[Post-Marketing (Inversión 1500€)] Nuevas ventas por mes: {tienda1.ventas_ultimos_3_meses} €"
)
print(f"[Post-Marketing] Nuevo total acumulado: {total_proyectado:.2f} €\n")


print("--- VENTAS DEL ÚLTIMO MES DE TODAS LAS TIENDAS ---")
for tienda in lista_tiendas:
    print(
        f"{tienda.nombre}: {tienda.ventas_ultimo_mes():.2f} € en el último mes."
    )

print("\n--- TIENDAS UBICADAS EN UNA 'AVENIDA' ---")
for tienda in lista_tiendas:
    if "Avenida" in tienda.direccion:
        print(f"• {tienda.nombre} ({tienda.direccion})")
```

## Ejercicio 2
En este ejercicio vamos a implementar una clase *Perro* en Python. La clase tiene las siguientes características:
* Cosas que sabemos seguro que tiene un perro
    * Tiene 4 patas
    * 2 orejas
    * 2 ojos
    * Una velocidad de 0. Por defecto, el perro está parado
* Cuando se inicialice:
    * El perro será de una determinada raza
    * Por defecto tendrá pelo "Marrón", a no ser que se diga lo contrario.
    * Por defecto no tendrá dueño, a no ser que se diga lo contrario.
    
* Dispondrá también de un método llamado andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro.
* Necesita otro método (parar), donde pondremos la velocidad a 0.
* Otro método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada.


Se pide:
* Implementa la clase *Perro*
* Crea un objeto de tipo *Perro*, sin dueño
* Comprueba que están bien todos sus atributos
* Prueba que ande, y comprueba su velocidad
* Páralo
* Documenta la clase *Perro*

```python
class Perro:
    """Clase para modelar el comportamiento y las características de un perro.

    Atributos de clase:
        patas (int): Número de patas del perro (4).
        orejas (int): Número de orejas del perro (2).
        ojos (int): Número de ojos del perro (2).
        velocidad (float): Velocidad inicial del perro en estado parado (0.0).
    """

    patas = 4
    orejas = 2
    ojos = 2
    velocidad = 0.0

    def __init__(
        self,
        raza: str,
        pelo: str = "Marrón",
        dueno: str | None = None,
    ):
        """Inicializa una nueva instancia de la clase Perro.

        Args:
            raza (str): Raza del perro.
            pelo (str, opcional): Color del pelo. Por defecto es 'Marrón'.
            dueno (str, opcional): Nombre del dueño si lo tiene. Por defecto es None.
        """
        self.raza = raza
        self.pelo = pelo
        self.dueno = dueno

    def andar(self, aumento_velocidad: float) -> float:
        """Incrementa la velocidad actual del perro.

        Args:
            aumento_velocidad (float): Valor que se suma a la velocidad actual.

        Returns:
            float: La nueva velocidad acumulada del perro.
        """
        self.velocidad += aumento_velocidad
        return self.velocidad

    def parar(self) -> float:
        """Detiene al perro, restableciendo su velocidad a 0.

        Returns:
            float: La velocidad tras detenerse (0.0).
        """
        self.velocidad = 0.0
        return self.velocidad

    def ladrar(self, texto: str) -> str:
        """Devuelve un ladrido acompañado de un mensaje.

        Args:
            texto (str): Texto o palabra a incluir tras el ladrido.

        Returns:
            str: Cadena formateada con "GUAU!" seguido del argumento recibido.
        """
        return f"GUAU!{texto}"

mi_perro = Perro(raza="Pastor Alemán")

print("--- COMPROBACIÓN DE ATRIBUTOS ---")
print(f"Patas (Atributo de clase): {mi_perro.patas}")
print(f"Orejas (Atributo de clase): {mi_perro.orejas}")
print(f"Ojos (Atributo de clase): {mi_perro.ojos}")
print(f"Velocidad inicial: {mi_perro.velocidad} km/h")
print(f"Raza: {mi_perro.raza}")
print(f"Color de pelo: {mi_perro.pelo}")
print(f"Dueño: {mi_perro.dueno}")



print("\n--- PRUEBA DEL MÉTODO ANDAR ---")
mi_perro.andar(5.5)
print(f"Velocidad tras andar por primera vez (+5.5): {mi_perro.velocidad} km/h")

mi_perro.andar(3.0)
print(f"Velocidad tras acelerar de nuevo (+3.0): {mi_perro.velocidad} km/h")


print("\n--- PRUEBA DEL MÉTODO PARAR Y LADRAR ---")
mi_perro.parar()
print(f"Velocidad tras ejecutar parar(): {mi_perro.velocidad} km/h")

ladrido = mi_perro.ladrar(" Dame una galleta")
print(f"Ladrido del perro: {ladrido}")
```
