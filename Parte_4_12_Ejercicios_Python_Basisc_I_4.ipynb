## Ejercicio 1
Para este ejercicio vamos a poner en práctica [las funciones built in](https://docs.python.org/3/library/functions.html).

1. Calcula el máximo de la lista: [4, 6, 8, -1]
2. Suma todos los elementos de la lista anterior
3. Redondea este float a 3 dígitos decimales: 63.451256965
4. Valor absoluto de: -74

```python
lista = [4, 6, 8, -1]
maximo = max(lista)
print("Máximo:", maximo)
suma = sum(lista)
print("Suma:", suma)
numero_float = 63.451256965
redondeado = round(numero_float, 3)
print("Redondeado:", redondeado)
absoluto = abs(-74)
print("Valor absoluto:", absoluto)
```

## Ejercicio 2
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."

1. Pásalo todo a mayúsculas
2. Pásalo todo a minúsculas
3. Solo la primera letra de cada palabra en mayúscula, el resto en minúscula
4. Crea una lista dividiéndolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`
6. Elimina las `a` minúsculas

```python
texto = "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."
mayusculas = texto.upper()
print("1. Mayúsculas:")
print(mayusculas)
print("-" * 50)

minusculas = texto.lower()
print("2. Minúsculas:")
print(minusculas)
print("-" * 50)


titulo = texto.title()
print("3. Primera letra de cada palabra en mayúscula:")
print(titulo)
print("-" * 50)

lista_palabras = texto.split(" ")
print("4. Lista dividida por espacios:")
print(lista_palabras)
print("-" * 50)

reemplazo_comas = texto.replace(",", ";")
print("5. Comas por puntos y comas:")
print(reemplazo_comas)
print("-" * 50)

sin_a = texto.replace("a", "")
print("6. Sin las 'a' minúsculas:")
print(sin_a)
```
